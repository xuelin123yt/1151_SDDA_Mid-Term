````python
# ============================================================
# 第 1 格：安裝套件、讀取金鑰、設定模型（請勿修改）
# ============================================================
!pip install -q groq gradio

import os
import re
import time
import sqlite3
import pandas as pd
from IPython.display import display
from groq import Groq
import gradio as gr

# 判斷目前是在 Colab 還是本地端 Jupyter
try:
    from google.colab import userdata
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

# 🔴 金鑰名稱一律為 GROQ_API_KEY；絕對不可以把金鑰寫在程式裡，也不可以 print 出來
if IN_COLAB:
    try:
        GROQ_API_KEY = userdata.get('GROQ_API_KEY')
    except Exception:
        print("❌ 讀不到金鑰：請到左側 🔑 新增名稱為 GROQ_API_KEY 的密鑰，並打開「筆記本存取權」")
        raise
else:
    GROQ_API_KEY = os.environ.get('GROQ_API_KEY')
    if not GROQ_API_KEY:
        raise RuntimeError("找不到環境變數 GROQ_API_KEY")

client = Groq(api_key=GROQ_API_KEY)

# 期中考指定模型；推理強度設為 low，可大幅減少 token 用量
# （唯一允許修改處：若 20b 額度用完，可改為 "openai/gpt-oss-120b"，並在規格書最後註明）
MODEL = "openai/gpt-oss-20b"
REASONING_EFFORT = "low"
print("✅ 環境：", "Colab" if IN_COLAB else "本地端", "｜模型：", MODEL)
````

````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
# 這一格做三件事：
# 1. 設定資料庫檔名與固定的常數（動作清單、活動清單、人數範圍、每人上限）
# 2. 定義 init_db()：刪掉舊表 → 建立兩張新表 → 放入初始資料
# 3. 執行 init_db()，確保每次「全部執行」都是乾淨的狀態

DB_PATH = "club.db"

ACTIONS = ["報名", "取消", "查詢", "統計"]          # AI 只能輸出這幾種動作（無法判斷時為「未知」）
EVENTS = ["迎新晚會", "程式工作坊", "籃球友誼賽"]     # AI 只能輸出這幾個活動（找不到時為「無」）
MIN_PEOPLE, MAX_PEOPLE = 1, 3                        # 一次報名的人數範圍（含攜伴）
MAX_ACTIVE_EVENTS = 2                                # 每位學生最多同時報名幾個活動（已取消的不計）

INITIAL_EVENTS = [
    ("迎新晚會", "2026-11-20", 6),
    ("程式工作坊", "2026-11-25", 4),
    ("籃球友誼賽", "2026-11-28", 3),
]


def get_conn():
    """建立資料庫連線（每次用完要記得關閉）"""
    return sqlite3.connect(DB_PATH)


def init_db():
    """重建兩張資料表並放入初始資料"""
    conn = get_conn()
    try:
        cur = conn.cursor()
        # 先刪「報名紀錄」再刪「活動」，因為報名紀錄會參考活動
        cur.execute("DROP TABLE IF EXISTS registrations")
        cur.execute("DROP TABLE IF EXISTS events")

        cur.execute("""
            CREATE TABLE events (
                event_id INTEGER PRIMARY KEY AUTOINCREMENT,
                name     TEXT NOT NULL UNIQUE,
                date     TEXT NOT NULL,
                capacity INTEGER NOT NULL
            )
        """)
        cur.execute("""
            CREATE TABLE registrations (
                reg_id    INTEGER PRIMARY KEY AUTOINCREMENT,
                student   TEXT NOT NULL,
                event_id  INTEGER NOT NULL,
                headcount INTEGER NOT NULL DEFAULT 1,
                status    TEXT NOT NULL DEFAULT '已報名',
                FOREIGN KEY (event_id) REFERENCES events(event_id)
            )
        """)

        cur.executemany(
            "INSERT INTO events (name, date, capacity) VALUES (?, ?, ?)",
            INITIAL_EVENTS,
        )
        conn.commit()
    finally:
        conn.close()


init_db()
print("✅ 資料庫已重建，初始資料已放入")
````

````python
# ============================================================
# 第 3 格：顯示資料表
# 提醒：定義 show_tables() 並呼叫一次
# ============================================================
# show_tables() 會把兩張資料表的內容用表格顯示出來，方便檢查資料庫有沒有改對。

def show_tables():
    """顯示 events 與 registrations 兩張表的全部內容"""
    conn = get_conn()
    try:
        events_df = pd.read_sql_query("SELECT * FROM events", conn)
        regs_df = pd.read_sql_query("SELECT * FROM registrations", conn)
    finally:
        conn.close()

    print("📅 events（活動）")
    display(events_df)
    print("📝 registrations（報名紀錄）")
    if regs_df.empty:
        print("（目前沒有任何報名紀錄）")
    else:
        display(regs_df)


show_tables()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 活動, 人數, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
# 這一格的流程：使用者一句話 → 呼叫 Groq AI → 拿到「動作,活動,人數」→
# 程式自己檢查 AI 的回覆（只接受允許清單裡的值）→ 回傳整理好的結果。

SYSTEM_PROMPT = """你是「系學會活動報名系統」的文字解讀器，只負責把使用者的一句話轉成一行固定格式。
輸出格式（只能輸出這一行，不要加任何說明、標點、空白行或引號）：
動作,活動,人數

規則：
1. 動作只能是：報名、取消、查詢、統計、未知。
   - 報名：想參加活動。 取消：想退出、不去了。 查詢：問某活動還剩多少名額。
   - 統計：想看「所有活動」的報名狀況或總覽。
   - 無法判斷或與活動無關時輸出：未知
2. 活動只能是：迎新晚會、程式工作坊、籃球友誼賽。使用者說的活動不在這三個之中，或沒有說活動，或同時提到多個活動時，活動輸出：無
   統計時活動輸出：無
3. 人數是「這次報名的總人數」，要把說話的人自己算進去。
   例如「我和兩個朋友」＝3、「我跟一個朋友」＝2、「我們五個人」＝5。沒有提到人數就輸出 1。
   人數只輸出阿拉伯數字，不要判斷人數合不合理（即使是 0 或 10 也照實輸出）。
4. 你不可以執行報名或取消，不可以回答問題，不可以編造不存在的活動。

範例：
我要報名迎新晚會 → 報名,迎新晚會,1
我和兩個朋友想報名程式工作坊 → 報名,程式工作坊,3
幫我取消籃球友誼賽 → 取消,籃球友誼賽,1
程式工作坊還剩幾個名額 → 查詢,程式工作坊,1
各活動報名狀況 → 統計,無,1
今天天氣怎麼樣 → 未知,無,1
"""

AI_CACHE = {}   # 同一句話只問一次 AI，結果暫存在這裡（節省額度）


def call_ai(text):
    """呼叫 Groq，回傳 AI 的原始文字。遇到 429（用量超過）等 30 秒再試，最多 3 次。"""
    for attempt in range(3):
        try:
            resp = client.chat.completions.create(
                model=MODEL,
                temperature=0,
                reasoning_effort=REASONING_EFFORT,
                max_completion_tokens=500,
                messages=[
                    {"role": "system", "content": SYSTEM_PROMPT},
                    {"role": "user", "content": text},
                ],
            )
            time.sleep(2)   # 每次呼叫後休息 2 秒，避免太快用完額度
            return (resp.choices[0].message.content or "").strip()
        except Exception as e:
            time.sleep(2)
            if "429" in str(e) or getattr(e, "status_code", None) == 429:
                print(f"⏳ 超過使用量，等待 30 秒後重試（第 {attempt + 1}/3 次）")
                time.sleep(30)
            else:
                print("⚠️ 呼叫 AI 失敗：", type(e).__name__)
                return ""
    return ""


def parse_ai_reply(raw):
    """不直接相信 AI：自己從回覆中找出允許清單裡的值。找不到就用預設值。"""
    # 動作：找「最早出現」的允許動作，找不到＝未知
    action, best = "未知", None
    for a in ACTIONS:
        pos = raw.find(a)
        if pos != -1 and (best is None or pos < best):
            action, best = a, pos

    # 活動：同樣找最早出現的允許活動，找不到＝無
    event, best = "無", None
    for ev in EVENTS:
        pos = raw.find(ev)
        if pos != -1 and (best is None or pos < best):
            event, best = ev, pos

    # 人數：優先看第 3 個欄位，用正規表示式找數字；找不到就當作 1
    parts = re.split(r"[,，]", raw)
    source = parts[2] if len(parts) >= 3 else raw
    m = re.search(r"\d+", source)
    count = int(m.group()) if m else 1
    return action, event, count


def interpret(text):
    """AI 解讀主函式：回傳 (動作, 活動, 人數, 原始回覆)"""
    key = text.strip()
    if key not in AI_CACHE:
        AI_CACHE[key] = call_ai(key)
    raw = AI_CACHE[key]
    action, event, count = parse_ai_reply(raw)
    return action, event, count, raw


print("✅ AI 解讀函式已定義")
````

````python
# ============================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ============================================================
# 格式：(使用者說的話, 預期動作, 預期對象, 預期數量)，"*" 表示不在意
AI_TESTS = [
    ("我要報名迎新晚會", "報名", "迎新晚會", 1),  # 1 正常
    ("幫我取消程式工作坊", "取消", "程式工作坊", "*"),  # 2 正常
    ("籃球友誼賽還剩幾個名額？", "查詢", "籃球友誼賽", "*"),  # 3 正常
    ("各活動報名狀況", "統計", "*", "*"),  # 4 正常（統計）
    ("我和兩個朋友想報名程式工作坊", "報名", "程式工作坊", 3),  # 5 正常（人數）
    ("我們五個人要報名迎新晚會", "報名", "迎新晚會", 5),  # 6 邊界（人數超過上限，由程式拒絕）
    ("我跟一個朋友要報名籃球友誼賽", "報名", "籃球友誼賽", 2),  # 7 邊界（攜伴）
    ("我要報名烤肉大會", "報名", "無", "*"),  # 8 錯誤（活動不存在）
    ("今天天氣怎麼樣", "未知", "無", "*"),  # 9 錯誤（無關的話）
    ("我想查詢", "查詢", "無", "*"),  # 10 邊界（沒說活動）
]


def run_ai_tests():
    passed = 0
    for i, (text, exp_action, exp_event, exp_count) in enumerate(AI_TESTS, start=1):
        action, event, count, raw = interpret(text)
        ok = (
            action == exp_action
            and (exp_event == "*" or event == exp_event)
            and (exp_count == "*" or count == exp_count)
        )
        passed += ok
        print(f"{i:>2}. {'✅' if ok else '❌'} 「{text}」")
        print(f"    AI 原始回覆：{raw!r}")
        print(f"    系統解讀：動作={action}、活動={event}、人數={count}"
              f"（預期：{exp_action}、{exp_event}、{exp_count}）")
    print(f"\n🎯 AI 解讀測試：通過 {passed} / {len(AI_TESTS)} 句")


run_ai_tests()
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
# 這一格定義系統的所有功能：報名、取消、查詢剩餘名額、統計，以及總入口 handle_request()。
# 名額一律用「人數加總」計算，只計算狀態為「已報名」的紀錄。

def _booked(cur, event_id):
    """某活動目前已報名的人數加總（已取消的不算）"""
    cur.execute(
        "SELECT COALESCE(SUM(headcount), 0) FROM registrations "
        "WHERE event_id = ? AND status = '已報名'",
        (event_id,),
    )
    row = cur.fetchone()
    return row[0] if row is not None else 0


def _find_event(cur, name):
    """用活動名稱找活動，回傳 (event_id, 日期, 名額)，找不到回傳 None"""
    cur.execute("SELECT event_id, date, capacity FROM events WHERE name = ?", (name,))
    return cur.fetchone()


def list_events_text():
    """列出目前所有活動（AI 看不懂時用）"""
    conn = get_conn()
    try:
        cur = conn.cursor()
        cur.execute("SELECT name, date FROM events ORDER BY date")
        rows = cur.fetchall()
    finally:
        conn.close()
    return "\n".join(f"・{name}（{date}）" for name, date in rows)


def register(student, event, n):
    """報名：檢查人數範圍 → 不可重複 → 每人上限 → 名額是否足夠 → 寫入"""
    if n < MIN_PEOPLE or n > MAX_PEOPLE:
        return f"❌ 報名人數需介於 {MIN_PEOPLE}～{MAX_PEOPLE} 人（含攜伴），您輸入的是 {n} 人。"

    conn = get_conn()
    try:
        cur = conn.cursor()
        ev = _find_event(cur, event)
        if ev is None:
            return "❌ 查無此活動。"
        event_id, date, capacity = ev

        # 同一位學生不可重複報名同一個活動（已取消的不算）
        cur.execute(
            "SELECT reg_id FROM registrations "
            "WHERE student = ? AND event_id = ? AND status = '已報名'",
            (student, event_id),
        )
        if cur.fetchone() is not None:
            return f"⚠️ {student} 已經報名「{event}」，不能重複報名。"

        # 每位學生最多同時報名 2 個活動（已取消的不計）
        cur.execute(
            "SELECT COUNT(DISTINCT event_id) FROM registrations "
            "WHERE student = ? AND status = '已報名'",
            (student,),
        )
        row = cur.fetchone()
        active = row[0] if row is not None else 0
        if active >= MAX_ACTIVE_EVENTS:
            return f"❌ {student} 已報名 {active} 個活動，達到每人上限（{MAX_ACTIVE_EVENTS} 個），請先取消其他活動。"

        # 名額是否足夠（用人數加總計算）
        remaining = capacity - _booked(cur, event_id)
        if n > remaining:
            return f"❌ 名額不足：「{event}」只剩 {remaining} 個名額，無法報名 {n} 人。"

        cur.execute(
            "INSERT INTO registrations (student, event_id, headcount, status) VALUES (?, ?, ?, '已報名')",
            (student, event_id, n),
        )
        conn.commit()
        return f"✅ 報名成功！{student} 報名「{event}」（{date}）共 {n} 人，剩餘名額 {remaining - n}。"
    finally:
        conn.close()


def cancel(student, event):
    """取消：把「已報名」的紀錄改成「已取消」，名額自動釋出"""
    conn = get_conn()
    try:
        cur = conn.cursor()
        ev = _find_event(cur, event)
        if ev is None:
            return "❌ 查無此活動。"
        event_id, date, capacity = ev

        cur.execute(
            "SELECT reg_id, headcount FROM registrations "
            "WHERE student = ? AND event_id = ? AND status = '已報名'",
            (student, event_id),
        )
        row = cur.fetchone()
        if row is None:
            return f"⚠️ 查無紀錄：{student} 沒有報名「{event}」，無法取消。"
        reg_id, headcount = row

        cur.execute("UPDATE registrations SET status = '已取消' WHERE reg_id = ?", (reg_id,))
        conn.commit()
        remaining = capacity - _booked(cur, event_id)
        return f"✅ 取消成功！已釋出 {headcount} 個名額，「{event}」剩餘名額 {remaining}。"
    finally:
        conn.close()


def query_remaining(event):
    """查詢：回覆某活動的剩餘名額（不需要姓名）"""
    conn = get_conn()
    try:
        cur = conn.cursor()
        ev = _find_event(cur, event)
        if ev is None:
            return "❌ 查無此活動。"
        event_id, date, capacity = ev
        remaining = capacity - _booked(cur, event_id)
        return f"🔎「{event}」（{date}）名額 {capacity} 人，剩餘名額 {remaining} 人。"
    finally:
        conn.close()


def get_stats():
    """統計：JOIN + GROUP BY 列出每個活動的日期、名額、已報名人數、剩餘名額。
    用 LEFT JOIN，而且「已報名」條件放在 ON 裡，這樣沒有人報名的活動也會列出來。"""
    conn = get_conn()
    try:
        cur = conn.cursor()
        cur.execute("""
            SELECT e.name, e.date, e.capacity, COALESCE(SUM(r.headcount), 0) AS booked
            FROM events AS e
            LEFT JOIN registrations AS r
                   ON r.event_id = e.event_id AND r.status = '已報名'
            GROUP BY e.event_id, e.name, e.date, e.capacity
            ORDER BY e.date
        """)
        rows = cur.fetchall()
    finally:
        conn.close()

    lines = ["📊 各活動報名狀況"]
    for name, date, capacity, booked in rows:
        lines.append(f"・{name}｜{date}｜名額 {capacity}｜已報名 {booked}｜剩餘 {capacity - booked}")
    return "\n".join(lines)


def unknown_reply():
    return "🤔 我看不懂您的需求，或是沒有這個活動。目前有以下活動：\n" + list_events_text()


def handle_request(text, student):
    """總入口：解讀一句話 → 依動作呼叫對應功能 → 回傳給使用者的文字"""
    text = (text or "").strip()
    student = (student or "").strip()
    if not text:
        return "請輸入您的需求，例如：我要報名迎新晚會。"

    action, event, n, raw = interpret(text)

    if action == "統計":
        return get_stats()
    if action == "未知" or event == "無":
        return unknown_reply()
    if action == "查詢":                      # 查詢不需要姓名
        return query_remaining(event)

    # 報名與取消需要姓名
    if not student:
        return "⚠️ 報名或取消需要輸入姓名。"
    if action == "報名":
        return register(student, event, n)
    if action == "取消":
        return cancel(student, event)
    return unknown_reply()


print("✅ 功能函式已定義")
````

````python
# ============================================================
# 第 7 格：主流程與流程測試
# 提醒：流程測試 8～10 步；最後呼叫 show_tables()
# ============================================================
# 流程測試：每一步用 handle_request() 跑一次完整流程，
# 並檢查系統回應有沒有包含「預期關鍵字」。每次都先重建資料庫，確保結果可重現。
init_db()

# 格式：(姓名, 使用者說的話, 預期回應中要出現的關鍵字)
FLOW_TESTS = [
    ("小明", "我要報名迎新晚會",             "報名成功"),     # 1 正常報名
    ("小明", "我要報名迎新晚會",             "不能重複報名"), # 2 重複報名被拒
    ("小明", "我和兩個朋友想報名程式工作坊", "報名成功"),     # 3 攜伴 3 人（剩 1 個名額）
    ("小明", "我跟一個朋友要報名籃球友誼賽", "每人上限"),     # 4 已有 2 個活動，被拒
    ("小美", "我跟一個朋友要報名程式工作坊", "名額不足"),     # 5 要 2 人但只剩 1 個名額
    ("小美", "我們五個人要報名迎新晚會",     "報名人數需介於"),# 6 人數超過 3 人，被拒
    ("小明", "幫我取消程式工作坊",           "取消成功"),     # 7 取消並釋出 3 個名額
    ("小華", "幫我取消程式工作坊",           "查無紀錄"),     # 8 沒報名過，查不到紀錄
    ("小華", "籃球友誼賽還剩幾個名額？",     "剩餘名額 3"),   # 9 查詢（不需要姓名）
    ("小華", "各活動報名狀況",               "各活動報名狀況"),# 10 統計（JOIN + GROUP BY）
]

flow_passed = 0
for i, (student, text, keyword) in enumerate(FLOW_TESTS, start=1):
    reply = handle_request(text, student)
    ok = keyword in reply
    flow_passed += ok
    print(f"【第 {i} 步】{'✅' if ok else '❌'} {student}：{text}")
    print(f"  系統回應：{reply}\n")

print(f"🎯 流程測試：通過 {flow_passed} / {len(FLOW_TESTS)} 步\n")
show_tables()
````

````python
# ============================================================
# 第 8 格：Gradio 網頁
# 提醒：if IN_COLAB: demo.launch(share=True)
# ============================================================
# 建立網頁：兩個輸入框（需求、姓名）→ 一個輸出框（系統回應）。

def web_handler(text, student):
    return handle_request(text, student)


demo = gr.Interface(
    fn=web_handler,
    inputs=[
        gr.Textbox(label="您的需求", placeholder="例如：我要報名迎新晚會"),
        gr.Textbox(label="姓名（報名、取消時必填；查詢可留空）", placeholder="例如：小明"),
    ],
    outputs=gr.Textbox(label="系統回應", lines=8),
    title="🎉 系學會活動報名系統",
    description="用一句話完成報名、取消、查詢剩餘名額，也可以輸入「各活動報名狀況」查看統計。",
)

if IN_COLAB:
    demo.launch(share=True)
else:
    print("目前在本地端：如要開啟網頁，請執行 demo.launch()")
````
